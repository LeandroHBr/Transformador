```python
import cmath
```

```python
# ensaio a vasio

Volt_v = 240 
Amp_v = 7.133
Pot_v = 400
```

```python
# obtendo a Resistência do núcle V^2/P
Rc = (Volt_v**2)/Pot_v;
print(Rc)

# desasagem entre a potência ativa e apartente S * coso = P
S = Volt_v*Amp_v
ang = cmath.acos(Pot_v/S)

# obtendo a potência Reativa e calculando XM = V^2/Q

Q = S * cmath.sin(ang)
XM = (Volt_v ** 2)/Q;
print(XM)
```
